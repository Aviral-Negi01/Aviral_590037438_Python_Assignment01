```python
text = '   Never Odd or Even  '
stripped_text = text.strip()
lower_text = stripped_text.lower()
normalized_text = lower_text.replace(' ','')

is_palindrome = normalized_text == normalized_text[::-1]

print("Given Text:", text) #'   Never Odd or Even  '
print("After strip():", stripped_text) # 'Never Odd or Even'
print("After lower():", lower_text) # 'never odd or even'
print("Normalized Text:", normalized_text) #'neveroddoreven'
print("Is palidrome:", is_palindrome)# True

text = 'Python'
stripped_text = text.strip()
lower_text = stripped_text.lower()
normalized_text = lower_text.replace(' ','')

is_palindrome = normalized_text == normalized_text[::-1]

print("Given Text:", text)  
print("After strip():", stripped_text) 
print("After lower():", lower_text) 
print("Normalized Text:", normalized_text) 
print("Is palidrome:", is_palindrome)
```
