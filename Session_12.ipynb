''' A file mode tells Python what kind of access the program is requesting. The common text modes are r for reading, w for writing and replacing, a for appending, and x for creating a new file only when the path does not already exist. The b flag selects binary data, t selects text data, and + adds both reading and writing to a base mode. open() returns a file object whose read(), write(), readline(), readlines(), and close() methods control the stream.'''

```python
mode = 'r'
print("Mode:", mode)
print("Can Read:", "r" in mode or "+" in mode)
print("Can Write:", any (flag in mode for flag in ("w", "a", "x", "+")))
```
