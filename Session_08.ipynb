```python

transaction = ('withdrawal', 25000)

match transaction:
    case ('deposit', amount) if amount > 0:
        result = 'Deposit accepted'
    case ('withdrawal', amount) if amount > 20000:
        result = 'Withdrawal flagged for review'
    case ('withdrawal', amount) if amount > 0:
        result = 'Withdrawal accepted'
    case _:
        result = 'Unsupported record'


print("Transaction:", transaction)
print("Matched result:", result)

transaction = ('withdrawal', 5000)



match transaction:
    case ('deposit', amount) if amount > 0:
        result = 'Deposit accepted'
    case ('withdrawal', amount) if amount > 20000:
        result = 'Withdrawal flagged for review'
    case ('withdrawal', amount) if amount > 0:
        result = 'Withdrawal accepted'
    case _:
        result = 'Unsupported record'

print("Transaction:", transaction)
print("Matched result:", result)
```
