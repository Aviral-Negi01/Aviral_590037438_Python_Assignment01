```python
principal = 50000.0
annual_rate = 0.08
years = 2
final_ammount = principal*(1+annual_rate)**years #50000.0*(1.08)**2
growth_ammount = final_ammount - principal

print(f"Principal Ammount: {principal}, Annual Rate: {annual_rate}%, years:{years}")
print("Final Ammount:", round(final_ammount,1))
print("Growth Ammount:", round(growth_ammount,1))

# Second Run
principal = 50000.0
annual_rate = 0.0
years = 2
final_ammount = principal*(1+annual_rate)**years #50000.0*(1.08)**2
growth_ammount = final_ammount - principal

print(f"Principal Ammount: {principal}, Annual Rate: {annual_rate}%, years:{years}")
print("Final Ammount:", round(final_ammount,1))
print("Growth Ammount:", round(growth_ammount,1))
```
