```python
total_seconds = 7384
print(f"{total_seconds} seconds")

hours = total_seconds // 3600 
remaining_seconds = total_seconds % 3600

minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print(f"Complete Hours : {hours}")
print(f"Remaining Seconds : {remaining_seconds}")
print(f"Complete Minutes :{minutes}")
print(f"Remaining Seconds :{seconds}")
print(f"Final Answer: {hours} hours, {minutes} minutes, {seconds} seconds")
# Second Run 
total_seconds = 59
print(f"{total_seconds} seconds")

hours = total_seconds // 3600 
remaining_seconds = total_seconds % 3600

minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print(f"Complete Hours : {hours}")
print(f"Remaining Seconds : {remaining_seconds}")
print(f"Complete Minutes :{minutes}")
print(f"Remaining Seconds :{seconds}")
print(f"Final Answer: {hours} hours, {minutes} minutes, {seconds} seconds")
```
